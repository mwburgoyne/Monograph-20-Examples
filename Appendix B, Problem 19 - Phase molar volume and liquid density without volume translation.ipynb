# Problem 19

Estimate the liquid phase molar volume (without volume translation) and the liquid
density for the 500 psia / 280 °F flash of Problem 18.

The component properties and the PR `b` / volume-shift terms come from the shared
`pvt_utils` module. The translated phase molar volumes and the flashed
compositions are carried over from Problem 18 (Table B-30).

> *Note:* the molecular weight quoted for the liquid in the original problem
> statement is inconsistent with the flashed `xi`, so we recompute `Ml` directly
> from the composition below.

```python
import pvt_utils as pvt

# Carried over from Problem 18 (Table B-30, flash at 500 psia & 280 degF)
vl = 2.721    # ft3/lbm mol - liquid molar volume (volume-translated)
vv = 13.837   # ft3/lbm mol - vapour molar volume (volume-translated)
yi = [0.57114, 0.41253, 0.01633]
xi = [0.08588, 0.46349, 0.45064]

comps = pvt.get_components(['C1', 'nC4', 'C10'])
ncomps = len(comps)
```

```python
# Peng-Robinson b and the Peneloux volume shift c = s * b for each component
bi = [pvt.pr_b(c.Tc, c.Pc) for c in comps]
ci = [bi[i] * comps[i].s for i in range(ncomps)]

# Liquid molar volume WITHOUT volume translation = translated volume + shift
vl_eos = vl + sum(xi[i] * ci[i] for i in range(ncomps))
print('Liquid molar volume without volume translation:', round(vl_eos, 3), ' ft3/lbm mol')
```

```python
# Recompute the liquid molecular weight from the flashed composition
Ml = sum(xi[i] * comps[i].M for i in range(ncomps))
print('Liquid molecular weight:', round(Ml, 2), 'lbm/lbm mol')
print('Uncorrected Liquid density:', round(Ml/vl_eos, 2), 'lbm/ft3 =', round(Ml/vl_eos/62.4, 3),
      'sg (', round((141.5/(Ml/vl_eos/62.4)) - 131.5, 1), 'deg API)')
print('  Corrected Liquid density:', round(Ml/vl, 2), 'lbm/ft3 =', round(Ml/vl/62.4, 3),
      'sg (', round((141.5/(Ml/vl/62.4)) - 131.5, 1), 'deg API)')
```
